Pseudocode for Root Calculator

PRINT Equation Root Finder
PRINT Select the operation to perform:
DISPLAY Operations Menu(1.Quadratic, 2.Quartic, 3.Cubic)
INPUT choice from user
IF choice is 1, 2, or 3:
    INPUT coefficients
    CALCULATE roots 
    DISPLAY the results
ELSE 
DISPLAY "Invalid Selection"

```python
import numpy as np

print("--- Equation Root Finder ---")
print("Select the operation to perform:")
print("1. Quadratic (Ax\u00b2 + Bx + C = 0)")
print("2. Cubic (Ax\u00b3 + Bx\u00b2 + Cx + D = 0)")
print("3. Quartic (Ax\u00B4 + Bx\u00b3 + Cx\u00b2 + Dx + E = 0)")

choice = input("\nEnter your choice (1, 2, or 3): ")

# Selection Logic
if choice == '1':
    print("\nQuadratic Equation")
    a = float(input("Enter A: "))
    b = float(input("Enter B: "))
    c = float(input("Enter C: "))
    coeffs = [a, b, c]
    
elif choice == '2':
    print("\nCubic Equation")
    a = float(input("Enter A: "))
    b = float(input("Enter B: "))
    c = float(input("Enter C: "))
    d = float(input("Enter D: "))
    coeffs = [a, b, c, d]

elif choice == '3':
    print("\nQuartic Equation")
    a = float(input("Enter A: "))
    b = float(input("Enter B: "))
    c = float(input("Enter C: "))
    d = float(input("Enter D: "))
    e = float(input("Enter E: "))
    coeffs = [a, b, c, d, e]

else:
    print("Invalid selection. Please restart.")
    coeffs = None

# Calculate and Display Results
if coeffs:
    roots = np.roots(coeffs)
    print("\nThe roots are:")
    for r in roots:
        print(f"{r:.2f}")
```
