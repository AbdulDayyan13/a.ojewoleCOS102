Flowchart for Simple Interest


```python
def get_positive_number(prompt):
    while True:
        try:
            value = float(input(prompt))
            if value > 0:
                return value
            else:
                print("Error: Please enter a number greater than 0")
        except ValueError:
            print("Invalid input!Please enter a numerical value (e.g 500 0r 0.005).")

print("Simple Interest Calculator.")
print("Enter your values below:")
num1 = get_positive_number("Enter Principal Value:")
num2 = get_positive_number("Enter Rate Value:")
num3 = get_positive_number("Enter Time  Value:")

amount = num1 * (1 + (num2/100) * num3)
print ("Amount is:", amount)
```

Flowchart for Compound Interest


```python
def get_positive_number(prompt):
    while True:
        try:
            value = float(input(prompt))
            if value > 0:
                return value
            else:
                print("Error: Please enter a number greater than 0")
        except ValueError:
            print("Invalid input!Please enter a numerical value (e.g 500 0r 0.05).")

print("Compound Interest Calculator")
print("Enter your values below:")
principal = get_positive_number("Enter Principal Value:")
rate = get_positive_number("Enter Rate Value:")
time = get_positive_number("Enter Time Value:")
n = int(get_positive_number("Enter n value:"))

amount = principal * (1 + (rate/100/n))** (n*time)
print("Amount is:", round(amount,2))
```

Flowchart for Annuity Plan


```python
def get_positive_number(prompt):
    while True:
        try:
            value = float(input(prompt))
            if value > 0:
                return value
            else:
                print("Error: Please enter a number greater than 0")
        except ValueError:
            print("Invalid input!Please enter a numerical value (e.g 500 0r 0.05).")

print("Annuity Plan Calculator")
print("Enter your values below:")

pmt = get_positive_number("Enter PMT Value:")
r = get_positive_number("Enter Rate Value(in whole numbers e.g 7):")
n = int(get_positive_number("Enter Frequency(n) Value:"))
t = get_positive_number("Enter number of years:")

amount = pmt * (((1 + r/100/n) ** (n*t) - 1) / (r/100/n))
print("Amount is:", round(amount,2))
```
