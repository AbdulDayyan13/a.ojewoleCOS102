```python
import tkinter as tk
from tkinter import messagebox
import csv

def welcomeMessage(username, department, colleagues):
    # Create the success window
    window = tk.Toplevel(root)
    window.title("Employee Verified")
    window.geometry("400x300")

    tk.Label(window, text=f"Welcome, {username}!" ).pack()
    tk.Label(window, text=f"Department: {department}").pack()
    
    tk.Label(window, text="\nYour Colleagues:").pack()
    
    # Create a listbox to show other members of the department
    listbox = tk.Listbox(window, width=40)
    listbox.pack()
    
    for member in colleagues:
        if member != username: # Don't list the user as their own colleague
            listbox.insert(tk.END, member)

def submit():
    input_name = username_entry.get().strip()
    input_dept = dept_entry.get().strip()

    found = False
    colleagues = []
    
    try:
        with open('GIG-logistics.csv', mode='r') as file:
            reader = csv.DictReader(file)
            
            # First, check if the employee exists
            all_employees = list(reader) # Convert to list to read multiple times
            
            for row in all_employees:
                # Check for match (Case-insensitive)
                if row['Name'].lower() == input_name.lower() and row['Department'].lower() == input_dept.lower():
                    found = True
                    break
            
            if found:
                # If found, get all people in that same department
                for row in all_employees:
                    if row['Department'].lower() == input_dept.lower():
                        colleagues.append(row['Name'])
                
                welcomeMessage(input_name, input_dept, colleagues)
            else:
                messagebox.showerror("Verification Failed", "Employee does not exist in this department.")
                
    except FileNotFoundError:
        messagebox.showerror("Error", "CSV file not found. Please ensure GIG-logistics.csv is in the folder.")

# --- UI Setup ---
root = tk.Tk()
root.title("GIG Logistics Verification")
root.geometry("300x250")

tk.Label(root, text="Enter Full Name:").pack(pady=5)
username_entry = tk.Entry(root)
username_entry.pack()

tk.Label(root, text="Enter Department:").pack(pady=5)
dept_entry = tk.Entry(root)
dept_entry.pack()

tk.Button(root, text="Verify", command=submit, bg="blue", fg="white").pack(pady=20)

root.mainloop()
```
